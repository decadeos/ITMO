**Задание 1**

В этом задании буду исследовать на устойчивость по критерию Рауса-Гурвица 3 уравнения.

## Первое уравнение

$$ y^{\prime\prime\prime\prime} + 5y^{\prime\prime\prime} + 18y^{\prime\prime} + 34y^{\prime} + 20y = 0 $$

Для начала составлю характеристическое уравнение:

$$
λ^4 + 5\lambda^3 + 18\lambda^2 + 34\lambda + 20
$$

Первый критерий это что все кофициенты > 0. Тут это выполняется, переходим к следующему действию.

Составлю матрицу Гурвица. Она будет размером 4 на 4, так как в диффуре 4 - старшая степень производной.

$$
G =
\begin{pmatrix}
5 & 34 & 0 & 0 \\
1 & 18 & 20 & 0 \\
0 & 5 & 34 & 0 \\
0 & 1 & 18 & 20 \\
\end{pmatrix}
$$

Теперь нужно посчитать все главные диаганальные миноры:

$$\Delta_1 = 5 > 0 $$
$$\Delta_2 = \begin{vmatrix} 5 & 34 \\ 1 & 18 \end{vmatrix} = 5 \cdot 18 - 34 \cdot 1 = 56 > 0 $$
$$\Delta_3 = \begin{vmatrix} 5 & 34 & 0 \\ 1 & 18 & 20 \\ 0 & 5 & 34 \end{vmatrix} = 1404 > 0 $$
$$\Delta_4 = 20 \cdot \Delta_3 > 0$$

Ура! Все главные диаганальные миноры больше нуля.
$$
\boxed{Система~асимптотически~устойчива}
$$

## Второе уравнение

$$
y''''+7y'''+19y''+23y'+10y = 0
$$
Все коэффициенты положительны, перехожу к матрице гурвица:


$$
G =
\begin{pmatrix}
7 & 23 & 0 & 0 \\
1 & 19 & 10 & 0 \\
0 & 7 & 23 & 0 \\
0 & 7 & 19 & 10 \\
\end{pmatrix}
$$

Посчитаю миноры:
$$
\Delta_1 = \det \begin{pmatrix} 7 \end{pmatrix} = 7
$$

$$
\Delta_2 = \det \begin{pmatrix}
7 & 23 \\
1 & 19 \\
\end{pmatrix} = 7 \cdot 19 - 23 \cdot 1 = 133 - 23 = 110
$$

$$
\Delta_3 = \det \begin{pmatrix}
7 & 23 & 0 \\
1 & 19 & 10 \\
0 & 7 & 23 \\
\end{pmatrix}
$$

$$
\Delta_3 = 0 \cdot \det \begin{pmatrix} 23 & 0 \\ 19 & 10 \end{pmatrix} - 7 \cdot \det \begin{pmatrix} 7 & 0 \\ 1 & 10 \end{pmatrix} + 23 \cdot \det \begin{pmatrix} 7 & 23 \\ 1 & 19 \end{pmatrix}
$$
$$
\Delta_3 = -7 \cdot 70 + 23 \cdot 110 = -490 + 2530 = 2040
$$

$$
\Delta_4 = \det \begin{pmatrix}
7 & 23 & 0 & 0 \\
1 & 19 & 10 & 0 \\
0 & 7 & 23 & 0 \\
0 & 7 & 19 & 10 \\
\end{pmatrix} = 20400
$$

Ура! Все главные диаганальные миноры больше нуля.
$$
\boxed{Система~асимптотически~устойчива}
$$

## Третье уравнение


$$
y''''+11y'''+41y''+61y'+30y = 0
$$
Все коэффициенты положительны, перехожу к матрице гурвица:


$$
G =

11 & 61 & 0 & 0 \\

0 & 11 & 61 & 0 \\
0 & 1 & 41 & 30 \\
\end{pmatrix}
$$


\begin{align*}
\Delta_1 &= 11 \\

Ура! Все главные диаганальные миноры больше нуля.

\boxed{Система~асимптотически~устойчива}

**Задание 2**

В этом задании нужно определить устойчивость в зависимости от параметра $\alpha$.
$$
y'''+2y''+\alpha y'+3y=0
$$

Воспользуюсь той же схемой, то есть буду решать с помощью критерием Рауса-Гурвица. Составлю матрицу гурвица, так как тут все коэффициенты положительны:

$$G =
\begin{pmatrix}
2 & 3 & 0\\
1 & \alpha & 0 \\
0 & 2 & 3 \\
\end{pmatrix}
$$

Теперь начнем исследовать главные миноры этой матрицы. Все не нужно, только тот, что размером 2 на 2:

$$
\Delta_2 =
\begin{vmatrix}
2 & 3 \\
1 & \alpha
\end{vmatrix} = 2\alpha - 3
$$

Это должно быть больше нуля, чтобы была устойчивость. Значит нужно решить сложнейшее неравенство:
$$
2\alpha - 3 >0 \\
\boxed{\alpha > \frac{3}{2}}
$$

Значит, уравнение устойчиво, при $\alpha > \frac{3}{2}$.

**Задание 3**
В этом задании буду исследовать уравнения на устойчивость по критерию Михайлова
## Первое уравнение

$$
y'''+y=0
$$

Суть решения в том, что нужно в характеристическое уравнение сделать подстановку годографа михайлова и на комплексной плоскости смотреть, как поведет себя это все дело. Я решила напистать уод на вольфрам математике для этого:


```
charPoly = lambda^3 + 1;

mikhailov = charPoly /. lambda -> I*omega;

re = ComplexExpand[Re[mikhailov]];
im = ComplexExpand[Im[mikhailov]];

ParametricPlot[{re, im}, {omega, 0, 100},
 PlotLabel -> "Годограф Михайлова для y''' + y = 0",
 AxesLabel -> {"Re", "Im"}, PlotRange -> All,
 Epilog -> {Red, PointSize[Large], Point[{re, im} /. omega -> 0]}]

angleChange = (ArcTan[re, im] /. omega -> 100) - (ArcTan[re, im] /.
     omega -> 0);
If[Chop[angleChange - 3*Pi/2] == 0,
 Print["Система устойчива (угол поворота равен 3*Pi/2)."],
 Print["Система неустойчива."]]
```

Итак, вот такой вывод получился:

```
Система неустойчива.
```
